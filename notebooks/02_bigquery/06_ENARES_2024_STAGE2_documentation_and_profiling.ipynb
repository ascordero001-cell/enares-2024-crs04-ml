## 3. Issue #13 — Estrategia de particionamiento BigQuery

Genera `ENARES_2024_STAGE2_partitioning_strategy.md` en `docs/`.

Importante: en Stage 02 solo se documenta la estrategia. No se crean tablas particionadas ni variables derivadas.

````python
partitioning_markdown = f"""# ENARES 2024 CRS04 - Stage 2 Estrategia de Particionamiento

Fecha UTC: {RUN_UTC}

## Alcance de Stage 02

Stage 02 solo documenta esta estrategia. No crea tablas particionadas ni variables derivadas sobre la capa raw.

La implementación física se difiere a Stage 03 sobre capas `cleaned` o `analytical`.

## Razón metodológica

Crear una variable derivada como `fecha_encuesta_referencia` en Stage 02 violaría el límite metodológico de la fase, porque Stage 02 no crea variables derivadas.

Crear tablas `*_partitioned` dentro de la capa raw también contaminaría la preservación intacta de microdatos.

## Estrategia recomendada

- Clave lógica de particionamiento futura: `anio_encuesta`.
- Si solo existe un año, como 2024, particionar por DATE genera una sola partición y no aporta poda ni ahorro.
- Para un único año se recomienda considerar `CLUSTER BY` sobre columnas de consulta frecuente.
- Para múltiples años se recomienda `PARTITION BY RANGE_BUCKET(anio_encuesta, GENERATE_ARRAY(2020, 2030, 1))`.
- Cualquier casting debe usar `SAFE_CAST` para tolerar valores nulos o no numéricos.

## SQL de referencia para Stage 03

```sql
-- Ejecutar solo en Stage 03 sobre capa cleaned/analytical, no sobre raw.

CREATE OR REPLACE TABLE `PROJECT_ID.enares2024_crs04_cleaned.crs04_cap100`
PARTITION BY RANGE_BUCKET(anio_encuesta, GENERATE_ARRAY(2020, 2030, 1))
AS
SELECT *
FROM `PROJECT_ID.enares2024_crs04_cleaned.crs04_cap100_staging`;
```

## Justificación técnica

El particionamiento reduce bytes procesados cuando las consultas filtran por la columna de partición. Para evaluar costo y rendimiento se debe registrar `total_bytes_processed`, `total_bytes_billed`, `slot_millis` y `cache_hit`.

En Stage 02 la evidencia se limita a documentación, observabilidad de tablas raw y cierre de competencia.
"""

partitioning_output = f"{DOCS_DIR}/ENARES_2024_STAGE2_partitioning_strategy.md"

with open(partitioning_output, "w", encoding="utf-8") as f:
    f.write(partitioning_markdown)

print("Issue #13 document saved:", partitioning_output)
````

## 6. Issue #14 — Profiling de solo lectura

Genera perfiles HTML con `ydata-profiling`.

Para evitar problemas de memoria en Colab, se usa una muestra configurable por tabla. El profiling es de solo lectura porque solo ejecuta `SELECT` sobre BigQuery y guarda HTML local en Drive.

Si deseas perfilar todas las filas, cambia `PROFILE_LIMIT` a `None`.

````python
# ============================================================
# Configuración de profiling
# ============================================================

PROFILE_LIMIT = 5000  # cambiar a None para todas las filas, si Colab tiene memoria suficiente

profiling_log = []

for table_name in RAW_TABLES:
    print("\nProfiling table:", table_name)

    table_ref = f"{PROJECT_ID}.{RAW_DATASET}.{table_name}"

    rowcount_before = client.query(
        f"SELECT COUNT(*) AS n FROM `{table_ref}`"
    ).result().to_dataframe()["n"].iloc[0]

    # Verificar si existe anio_encuesta antes de filtrar.
    table = client.get_table(table_ref)
    schema_names = {field.name for field in table.schema}

    if "anio_encuesta" in schema_names:
        where_clause = "WHERE anio_encuesta = 2024"
    else:
        where_clause = ""

    limit_clause = f"LIMIT {PROFILE_LIMIT}" if PROFILE_LIMIT is not None else ""

    query = f'''
    SELECT *
    FROM `{table_ref}`
    {where_clause}
    {limit_clause}
    '''

    df = client.query(query).result().to_dataframe()

    if df.empty:
        raise ValueError(f"Profiling query returned 0 rows for {table_name}")

    profile = ProfileReport(
        df,
        title=f"ENARES 2024 CRS04 - {table_name}",
        minimal=True,
    )

    profile_output = f"{LOG_DIR}/ENARES_2024_STAGE2_profile_{table_name}.html"
    profile.to_file(profile_output)

    rowcount_after = client.query(
        f"SELECT COUNT(*) AS n FROM `{table_ref}`"
    ).result().to_dataframe()["n"].iloc[0]

    profiling_log.append({
        "table_name": table_name,
        "table_ref": table_ref,
        "profile_limit": PROFILE_LIMIT,
        "profiled_rows": len(df),
        "rowcount_before": rowcount_before,
        "rowcount_after": rowcount_after,
        "raw_data_unchanged": rowcount_before == rowcount_after,
        "profile_output": profile_output,
        "profile_exists": os.path.exists(profile_output),
        "profile_size_bytes": os.path.getsize(profile_output) if os.path.exists(profile_output) else None,
        "checked_at_utc": RUN_UTC,
    })

profiling_validation = pd.DataFrame(profiling_log)

profiling_validation_output = f"{LOG_DIR}/ENARES_2024_STAGE2_profiling_validation.csv"
profiling_validation.to_csv(profiling_validation_output, index=False)

display(profiling_validation)

if not profiling_validation["raw_data_unchanged"].all():
    raise AssertionError("Profiling no cumple modo solo lectura: cambiaron conteos.")

if not profiling_validation["profile_exists"].all():
    missing_profiles = profiling_validation.loc[~profiling_validation["profile_exists"], "table_name"].tolist()
    raise FileNotFoundError("Faltan perfiles HTML: " + ", ".join(missing_profiles))

print("Issue #14 profiling PASS")
print("Saved:", profiling_validation_output)
````
